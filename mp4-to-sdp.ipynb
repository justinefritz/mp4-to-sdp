# Libraries

```python
import subprocess
import imageio_ffmpeg
```

# Convert Function

```python
def generate_sdp_and_stream(input_mp4, output_sdp, video_port=5004, audio_port=5006):
    ffmpeg_bin = imageio_ffmpeg.get_ffmpeg_exe()

    # Define separate RTP destinations for Video and Audio
    video_dest = f"rtp://127.0.0.1:{video_port}"
    audio_dest = f"rtp://127.0.0.1:{audio_port}"

    # Generate SDP File mapping Video and Audio to distinct ports
    sdp_command = [
        ffmpeg_bin,
        "-i", input_mp4,
        "-map", "0:v:0", "-c:v", "copy", "-f", "rtp", video_dest,
        "-map", "0:a:0", "-c:a", "copy", "-f", "rtp", audio_dest,
        "-sdp_file", output_sdp
    ]

    print(f"Generating SDP file: {output_sdp}...")
    sdp_process = subprocess.Popen(sdp_command, stdout=subprocess.PIPE, stderr=subprocess.PIPE)
    
    try:
        sdp_process.wait(timeout=2)
    except subprocess.TimeoutExpired:
        sdp_process.kill()
        sdp_process.wait()

    print(f"SDP successfully generated at {output_sdp}")
```

# Runner

```python
INPUT_PATH = "Kavi Talkies Philippines.mp4"
OUTPUT_SDP_PATH = "Kavi Talkies Philippines.sdp"

generate_sdp_and_stream(INPUT_PATH, OUTPUT_SDP_PATH)
```
